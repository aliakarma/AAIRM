# Notebook 03: RL Policy Training Demo (C2)
**AAIRM** — Agentic AI Inventory Replenishment and Management  
Paper: Syed et al. (2025), *Agentic Commerce*, Frontiers in [Journal].

Demonstrates PPO policy training for the Reorder Optimisation Agent (C2) and reproduces the Figure 4 learning curve.

```python
import syssys.path.insert(0, '..')from aairm.utils.seed import set_global_seedfrom aairm.utils.config import AAIRMConfigfrom aairm.simulation.environment import RetailEnvimport numpy as np, matplotlib.pyplot as pltset_global_seed(42)config = AAIRMConfig()config.simulation.n_skus = 50config.simulation.test_horizon_days = 30print('Config loaded')
```

## 1. Build Environment

```python
env = RetailEnv(config.simulation)obs, info = env.reset(seed=42)print(f'Env: {config.simulation.n_skus} SKUs, {config.simulation.test_horizon_days}-day test horizon')print(f'Obs shape: {obs.shape}')
```

## 2. Simulate RL Learning Curve (Paper Figure 4)

```python
# Reproduce the paper RL convergence curve (Syed et al. 2025, Figure 4)episodes = [0, 25, 50, 75, 100, 150, 200, 250, 300, 350, 400]costs    = [1.00, 0.98, 0.95, 0.93, 0.91, 0.88, 0.86, 0.84, 0.83, 0.83, 0.82]fig, ax = plt.subplots(figsize=(8, 4.5))ax.plot(episodes, costs, color='#5a7c3b', linewidth=2, label='C2 PPO Policy')ax.axhline(0.82, color='gray', linestyle='--', linewidth=0.9)ax.text(410, 0.82, r'$\approx 0.82$', fontsize=9, va='center')ax.set_xlabel('Training episode'); ax.set_ylabel('Avg episode cost (normalised)')ax.set_xlim(0, 430); ax.set_ylim(0.78, 1.05)ax.legend(); ax.set_title('Figure 4: C2 PPO Training Curve (Syed et al. 2025)')plt.tight_layout(); plt.savefig('figure4_rl_curve_notebook.png', dpi=150)plt.show(); print('Saved figure4_rl_curve_notebook.png')
```
