# Notebook 01: Data Exploration
**AAIRM** — Agentic AI Inventory Replenishment and Management

Paper: Syed et al. (2025), *Agentic Commerce*, Frontiers in [Journal].

This notebook explores the AAIRM synthetic simulation dataset: SKU distribution, demand patterns, seasonality, and supplier catalog statistics.

```python
import sys, numpy as np, pandas as pd
sys.path.insert(0, "..")
from aairm.utils.seed import set_global_seed
from aairm.simulation.sku_catalog import SKUCatalog
from aairm.simulation.demand_generator import DemandGenerator
from aairm.simulation.supplier_simulator import SupplierSimulator
set_global_seed(42)
print("Imports OK")
```

## 1. SKU Catalog Statistics

```python
catalog = SKUCatalog(n_skus=1200, seed=42)
df = catalog.to_dataframe()
print(f"Total SKUs: {len(df)}")
print("\nCategory distribution:")
print(df["category"].value_counts())
print("\nPerishable vs non-perishable:")
print(df["is_perishable"].value_counts())
```

## 2. Demand Pattern Visualisation

```python
gen = DemandGenerator(catalog, n_days=730, seed=42)
import matplotlib.pyplot as plt
# Plot 30-day demand for one SKU per category
fig, axes = plt.subplots(1, 5, figsize=(18, 3))
for ax, cat in zip(axes, catalog.categories):
    sku = catalog.skus_by_category(cat)[0]
    history = gen.get_history(sku, 30, 365)
    ax.plot(history, color="steelblue", linewidth=1)
    ax.set_title(cat.replace("_"," ").title())
    ax.set_xlabel("Day")
fig.suptitle("30-day Demand Trajectories (one SKU per category)")
plt.tight_layout(); plt.show()
```

## 3. Supplier Catalog Statistics

```python
sup = SupplierSimulator(catalog, seed=42)
rows = []
for sku_id in catalog.sku_ids[:100]:
    rows.extend(sup.query_catalogue(sku_id))
sup_df = pd.DataFrame(rows)
print(f"Suppliers (sample 100 SKUs): {len(sup_df)}")
print("\nAvg reliability by country:")
print(sup_df.groupby("country")["reliability"].mean().sort_values(ascending=False))
```
